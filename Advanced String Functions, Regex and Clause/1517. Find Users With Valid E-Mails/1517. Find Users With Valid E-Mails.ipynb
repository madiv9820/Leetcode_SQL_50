### 📋 Your Task
You are given a table named **`Users`** containing information about registered users.

Each record contains:
- 🆔 **`user_id`** — Unique ID of the user.
- 👤 **`name`** — User's name.
- 📧 **`mail`** — The email address provided during registration.

Your goal is to find all users whose email addresses are valid.

#### ✅ What Makes an Email Valid?
A valid email consists of two parts:

**1️⃣ Prefix (Before @)** <br>
The prefix:
- **✅ Must begin with an English letter (`A-Z` or `a-z`)**
- After the first letter, it may contain:
    - 🔤 Letters (**`A-Z`**, **`a-z`**)
    - 🔢 Digits (**`0-9`**)
    - **`_`** (underscore)
    - **`.`** (period)
    - **`-`** (dash)

**2️⃣ Domain (After @)** <br>
The domain must be exactly
```
@leetcode.com
```
Nothing else is allowed.
- **✅ `@leetcode.com`**
- **❌ `@LeetCode.com`**
- **❌ `@gmail.com`**
- **❌ `@leetcode.co`**
- **❌ `@leetcode.com.in`**

#### 🎯 Return
Return ***all users whose email addresses satisfy every rule above.*** <br>
The result can be returned in **any order**.

````python
from typing import List

from pyspark.sql import SparkSession

from pyspark.sql.types import (
    StructField,
    StructType,
    StringType,
    IntegerType,
)

import pyspark.sql.functions as F
````

````python
# =============================================================================
# 🚀 Create a Spark Session
# -----------------------------------------------------------------------------
# A SparkSession is the entry point to working with Spark.
#
# It allows us to:
#   • Create DataFrames
#   • Execute SQL queries
#   • Read and write data
#   • Perform distributed data processing
#
# Every PySpark application begins by creating a SparkSession.
# =============================================================================
spark = (
    SparkSession
        .builder
        .appName("Find Users With Valid E-Mails")
        .getOrCreate()
)
````

````python
# =============================================================================
# 📦 Sample datasets for testing
# -----------------------------------------------------------------------------
# Each inner list represents the data for one test case.
#
# Test Case 1
#   • Email has an uppercase domain (.COM) ❌
#   • Should NOT be considered valid.
#
# Test Case 2
#   • Contains a mixture of valid and invalid email addresses.
#   • Helps verify different validation scenarios.
# =============================================================================
user_test_cases: List = [
    [
        [1, "Winston", "winston@leetcode.COM"],
    ],
    [
        [1, "Winston", "winston@leetcode.com"],
        [2, "Jonathan", "jonathanisgreat"],
        [3, "Annabelle", "bella-@leetcode.com"],
        [4, "Sally", "sally.come@leetcode.com"],
        [5, "Marwan", "quarz#2020@leetcode.com"],
        [6, "David", "david69@gmail.com"],
        [7, "Shapiro", ".shapo@leetcode.com"],
    ],
]


# =============================================================================
# 🏗️ Schema definition
# -----------------------------------------------------------------------------
# Define the structure of the Users table.
#
# Columns:
#   • user_id → Unique identifier of the user
#   • name    → User's name
#   • mail    → User's registered email address
# =============================================================================
users_table_schema: StructType = StructType(
    [
        StructField("user_id", IntegerType(), True),
        StructField("name", StringType(), True),
        StructField("mail", StringType(), True),
    ]
)
````

---
### 🚀 Approach: Regex Pattern Matching
#### 💡 Intuition
Validating emails manually would require checking multiple conditions separately — whether the email starts with a letter, contains only allowed characters, and ends with the correct domain. 📧

Instead, we use a **Regular Expression (Regex)** that describes the complete structure of a valid LeetCode email. The regex automatically filters out invalid emails by checking whether the entire email matches the required format.

#### 🔍 Steps
1. Define a regex pattern:
```
^[a-zA-Z][a-zA-Z0-9._-]*@leetcode\.com$
```

2. Apply the pattern on the **`mail`** column using:
    - **`REGEXP`** in Spark SQL
    - **`rlike()`** in PySpark DataFrame API

3. Keep only those users whose emails match the pattern.

#### 🧩 Regex Breakdown
- **`^[a-zA-Z]`** → Email must start with a letter.
- **`[a-zA-Z0-9._-]*`** → Remaining prefix can contain letters, digits, **`.`**, **`_`**, and **`-`**.
- **`@leetcode\.com`** → Domain must be exactly **`@leetcode.com`**.
- **`$`** → Ensures no extra characters exist after the domain.

#### 📝 Pseudocode
```
Create email validation regex

For each user:
    Check if mail matches regex

    If valid:
        Add user to result

Return valid users
```

#### ⏱️ Complexity Analysis
- **Time Complexity: `O(n × m)`**
    - Each of the **`n`** emails is checked, and each check takes up to **`m`** characters.
- **Space Complexity: `O(1)`**
    - No additional data structures are used.
---

````python
# =============================================================================
# 📧 Regular Expression for validating email addresses
# -----------------------------------------------------------------------------
# Rules:
#
#   ^                         → Start of string
#
#   [a-zA-Z]                 → First character must be a letter
#
#   [a-zA-Z0-9._-]*          → Remaining prefix may contain:
#                                • Letters
#                                • Digits
#                                • Period (.)
#                                • Underscore (_)
#                                • Dash (-)
#
#   @leetcode\.com          → Domain must be exactly:
#                               @leetcode.com
#
#   $                        → End of string
# =============================================================================
valid_email_pattern = r"^[a-zA-Z][a-zA-Z0-9._-]*@leetcode\.com$"


# =============================================================================
# 🚀 Execute every test case
# =============================================================================
for test_case_number, current_users in enumerate(user_test_cases, start=1):

    print(f"\n{'=' * 70}")
    print(f"🧪 Test Case {test_case_number}")
    print(f"{'=' * 70}")

    # -------------------------------------------------------------------------
    # 📥 Create a DataFrame for the current test case
    # -------------------------------------------------------------------------
    users_dataframe = spark.createDataFrame(
        current_users,
        users_table_schema,
    )

    print("📋 Input Data")
    users_dataframe.show(truncate=False)

    # -------------------------------------------------------------------------
    # 🗂️ Register the DataFrame as a temporary SQL table.
    #
    # This allows us to execute Spark SQL queries directly on it.
    # -------------------------------------------------------------------------
    users_dataframe.createOrReplaceTempView("Users")

    # -------------------------------------------------------------------------
    # 🔍 SQL Solution
    #
    # Select only those users whose email addresses satisfy the required
    # regular expression.
    # -------------------------------------------------------------------------
    sql_query = f"""
        SELECT  user_id, name, mail
        FROM    Users
        WHERE   mail REGEXP '{valid_email_pattern}'
    """

    sql_result_dataframe = spark.sql(sql_query)

    print("🟦 SQL Result")
    sql_result_dataframe.show(truncate=False)

    # -------------------------------------------------------------------------
    # ⚡ PySpark DataFrame Solution
    #
    # filter()
    #     Keep only rows whose email matches the regular expression.
    #
    # rlike()
    #     Performs regex matching on the specified column.
    #
    # select()
    #     Return only the required columns.
    # -------------------------------------------------------------------------
    dataframe_result = (
        users_dataframe
            .filter(F.col("mail").rlike(valid_email_pattern))
            .select("user_id", "name", "mail")
    )

    print("🟨 PySpark DataFrame Result")
    dataframe_result.show(truncate=False)

    print()
````
