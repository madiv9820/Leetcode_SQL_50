February has always been a special month for **LeetMart**. With customers preparing for new projects, gifting seasons, and fresh starts, the company wants to understand which products truly became customer favourites during this period.

The sales analytics team has been asked to prepare a February performance report. However, they don't want to look at every product — only the products that crossed an important milestone: **selling at least 100 units during February 2020**.

Your task is to help the analytics team ***identify these high-performing products by combining product information with order records and calculating the total number of units sold during the month***.

### 📦 Problem Explanation
You are given two tables:

**🏷️ Products Table**

This table contains information about all products available in the store.

Each product has:
- A unique **`product_id`**
- A **`product_name`**
- A **`product_category`**

Example:
| product_id | product_name          | product_category |
| ---------- | --------------------- | ---------------- |
| 1          | Leetcode Solutions    | Book             |
| 2          | Jewels of Stringology | Book             |
| 3          | HP                    | Laptop           |

**🧾 Orders Table**

This table stores every order transaction.

Each record contains:
- The product that was ordered
- The date when it was ordered
- The number of units purchased

A product can appear multiple times because customers may place multiple orders on different days.

Example:
| product_id | order_date | unit |
| ---------- | ---------- | ---- |
| 1          | 2020-02-05 | 60   |
| 1          | 2020-02-10 | 70   |

The total February sales for this product are:
```
60 + 70 = 130 units
```

#### 🎯 Your Goal

Find all products that:
1. Were ordered during **February 2020**
2. Had a combined order quantity of **at least 100 units**
3. Return:
    - The product name
    - The total number of units ordered

The result can be returned in any order.

### 🟣 Approach 01: Pattern Matching on Date (`LIKE`)
#### 💡 Idea
This approach identifies February 2020 orders by treating the date as a pattern and matching dates that start with **`"2020-02"`**.

#### 🔍 Steps
1. Filter the **`Orders`** table to keep only records where **`order_date`** matches the February 2020 pattern.
2. Join the filtered orders with the **`Products`** table using **`product_id`**.
3. Group orders by **`product_name`**.
4. Calculate the total units ordered for each product.
5. Keep only products with total units greater than or equal to 100.

#### ⚡ Complexity
- **Time Complexity: `O(N + M)`** <br>
    **`N`** = number of products <br>
    **`M`** = number of orders
- **Space Complexity: `O(N)`** for intermediate aggregation results

#### ✅ Pros
- Simple and intuitive.
- Easy to understand when working with formatted date strings.

#### ⚠️ Cons
- Relies on date formatting.
- Less suitable when dates are stored in different formats.
- Pattern matching can be less efficient compared to native date operations.
---

````python
def approach_01_filter_using_pattern_matching(Products: DataFrame, Orders: DataFrame) -> Tuple[DataFrame]:
    """
    Finds products ordered in February 2020 with total units >= 100.

    Two implementations are demonstrated:
        1. Spark SQL approach
        2. PySpark DataFrame API approach

    Returns:
        Tuple containing:
            - SQL result DataFrame
            - DataFrame API result DataFrame
    """
    # 🗂️ Register both DataFrames as temporary SQL views so they can be queried using Spark SQL.
    # This allows us to query them using Spark SQL syntax.
    Products.createOrReplaceTempView('Products')
    Orders.createOrReplaceTempView('Orders')

    # 📝 SQL Approach:
    # 1. Filter only February 2020 orders.
    # 2. Join Products and Orders using product_id.
    # 3. Calculate total units purchased per product.
    # 4. Keep only products with >= 100 units.
    sql_query = '''
        WITH feb_orders AS (
            SELECT  *
            FROM    Orders
            WHERE   order_date LIKE '2020-02%'
        ),
        feb_products AS (
            SELECT      p.product_name 
                    ,   SUM(o.unit)     AS unit
            FROM        Products    p
            INNER JOIN  feb_orders  o
            ON          p.product_id = o.product_id
            GROUP BY    p.product_name
        )
        SELECT  *
        FROM    feb_products
        WHERE   unit >= 100
    '''

    sql_result_dataframe: DataFrame   = spark.sql(sql_query)
    
    # 🟨 PySpark DataFrame API Approach
    dataframe_result: DataFrame   = (
        Orders
            # 📅 Select only February orders
            .filter(F.col('order_date').like('2020-02%'))
            # 🔗 Combine orders with product details
            .join(Products, Orders['product_id'] == Products['product_id'], 'inner')
            # 📊 Calculate total units purchased per product
            .groupBy(F.col('product_name'))
            .agg(F.sum(F.col('unit')).alias('unit'))
            # 🎯 Keep only products meeting requirement
            .filter(F.col('unit') >= 100)
            # 🧹 Select required output columns
            .select('product_name', 'unit')
    )

    # ✅ Return both SQL and PySpark DataFrame solutions.
    return sql_result_dataframe, dataframe_result
````

### 🟢 Approach 02: Date Range Filtering (`BETWEEN`)
#### 💡 Idea
This approach filters orders by checking whether the order date falls within the start and end dates of February 2020.

#### 🔍 Steps
1. Select orders where: **`2020-02-01 <= order_date <= 2020-02-29`**
2. Join matching orders with product information.
3. Group orders by product name.
4. Sum all units ordered for each product.
5. Return products with at least 100 units sold.

#### ⚡ Complexity
- **Time Complexity: `O(N + M)`**
- **Space Complexity: `O(N)`** for grouped results

#### ✅ Pros
- Works directly with date values.
- Clear and readable.
- Efficient for filtering a specific time period.
- Works well with date indexes and partition pruning in large datasets.

#### ⚠️ Cons
- Requires knowing the exact start and end dates.
- Special handling may be needed for months with different numbers of days.
---

````python
def approach_02_filter_using_between_dates(Products: DataFrame, Orders: DataFrame) -> Tuple[DataFrame]:
    """
    Finds products ordered in February 2020 with total units >= 100.

    Two implementations are demonstrated:
        1. Spark SQL approach
        2. PySpark DataFrame API approach

    Returns:
        Tuple containing:
            - SQL result DataFrame
            - DataFrame API result DataFrame
    """
    # 🗂️ Register both DataFrames as temporary SQL views so they can be queried using Spark SQL.
    # This allows us to query them using Spark SQL syntax.
    Products.createOrReplaceTempView('Products')
    Orders.createOrReplaceTempView('Orders')

    # 📝 SQL Approach:
    # 1. Join Products and Orders using product_id.
    # 2. Keep only the orders placed during February 2020.
    # 3. Group records by product name.
    # 4. Calculate the total units sold for each product.
    # 5. Return only those products whose total sales are at least 100 units.
    sql_query = '''
        SELECT 	p.product_name, SUM(o.unit) AS unit
        FROM 		Products 	p
        INNER JOIN 	Orders 		o
        ON 			p.product_id = o.product_id
        WHERE 		o.order_date BETWEEN '2020-02-01' AND '2020-02-29'
        GROUP BY 	p.product_name
        HAVING 		SUM(o.unit) >= 100
    '''

    # 🚀 Execute the SQL query.
    sql_result_dataframe: DataFrame = spark.sql(sql_query)

    # 🟨 PySpark DataFrame API Approach
    dataframe_result: DataFrame = (
        Products
            # 🔗 Join products with their corresponding orders.
            .join(Orders, Products['product_id'] == Orders['product_id'], 'inner')
            # 📅 Keep only orders placed during February 2020.
            .filter(F.col('order_date').between('2020-02-01', '2020-02-29'))
            # 📦 Group all February orders by product name.
            .groupBy(F.col('product_name'))
            # ➕ Calculate the total number of units sold.
            .agg(F.sum(F.col('unit')).alias('unit'))
            # 🎯 Keep only products with at least 100 units sold.
            .filter(F.col('unit') >= 100)
            # 📋 Return only the required columns.
            .select('product_name', 'unit')
    )

    # ✅ Return both SQL and PySpark DataFrame solutions.
    return sql_result_dataframe, dataframe_result
````

### 🟠 Approach 03: Built-in Date Functions (YEAR + MONTH)
#### 💡 Idea
This approach uses Spark's built-in date functions to extract the year and month from each order date and filter matching records.

#### 🔍 Steps
1. Extract the year from order_date and keep only records where: **`YEAR(order_date) = 2020`**
2. Extract the month and keep only: **`MONTH(order_date) = 2`**
3. Join filtered orders with products.
4. Group by product name and calculate total units.
5. Keep products where total units are at least 100.

#### ⚡ Complexity
- **Time Complexity: `O(N + M)`**
- **Space Complexity: `O(N)`** for aggregation

#### ✅ Pros
- More readable and expressive.
- Does not depend on date string formatting.
- Flexible for dynamic date-based analysis.

#### ⚠️ Cons
- Applying functions on columns may prevent some database optimizations, such as partition pruning or index usage, depending on the execution engine.
---

````python
def approach_03_filter_using_year_and_month_functions(Products: DataFrame, Orders: DataFrame) -> Tuple[DataFrame]:
    """
    Finds products ordered in February 2020 with total units >= 100.

    Two implementations are demonstrated:
        1. Spark SQL approach
        2. PySpark DataFrame API approach

    Returns:
        Tuple containing:
            - SQL result DataFrame
            - DataFrame API result DataFrame
    """
    # 🗂️ Register both DataFrames as temporary SQL views so they can be queried using Spark SQL.
    # This allows us to query them using Spark SQL syntax.
    Products.createOrReplaceTempView('Products')
    Orders.createOrReplaceTempView('Orders')

    # 📝 SQL Approach:
    # 1. Join the Products and Orders tables using product_id.
    # 2. Filter only the orders placed during February 2020 by using
    #    the built-in YEAR() and MONTH() functions.
    # 3. Group records by product name.
    # 4. Calculate the total number of units ordered for each product.
    # 5. Return only those products whose total sales are at least 100 units.
    sql_query = '''
        SELECT 	p.product_name, SUM(o.unit) AS unit
        FROM 		Products 	p
        INNER JOIN 	Orders 		o
        ON 			p.product_id = o.product_id
        WHERE 		YEAR(o.order_date) 	= 2020 	AND
        			MONTH(o.order_date) = 2
        GROUP BY 	p.product_name
        HAVING 		SUM(o.unit) >= 100
    '''

    # 🚀 Execute the SQL query.
    sql_result_dataframe: DataFrame = spark.sql(sql_query)

    # 🟨 PySpark DataFrame API Approach
    dataframe_result: DataFrame = (
        Products
            # 🔗 Join products with their corresponding orders.
            .join(Orders, Products['product_id'] == Orders['product_id'], 'inner')
            # 📅 Keep only orders placed during February 2020.
            .filter(
                (F.year(F.col('order_date')) 	== 2020) & 
                (F.month(F.col('order_date')) 	== 2)
            )
            # 📦 Group all February orders by product name.
            .groupBy(F.col('product_name'))
            # ➕ Calculate the total number of units sold.
            .agg(F.sum(F.col('unit')).alias('unit'))
            # 🎯 Keep only products with at least 100 units sold.
            .filter(F.col('unit') >= 100)
            # 📋 Return only the required columns.
            .select('product_name', 'unit')
    )

    # ✅ Return both SQL and PySpark DataFrame solutions.
    return sql_result_dataframe, dataframe_result
````

````python
# 🧪 Store all approaches
approaches: List[Tuple[str, callable]] = [
    ("🟣 Pattern Matching", approach_01_filter_using_pattern_matching),
    ("🟢 BETWEEN Dates", approach_02_filter_using_between_dates),
    ("🟠 YEAR & MONTH Functions", approach_03_filter_using_year_and_month_functions),
]


# 📊 Store final comparison results
overall_summary = []


# 🧪 Execute all test cases
for test_case_number, unit_test_title in enumerate(unit_test_cases, start=1):

    test_case_result = {
        "Test Case": f"{unit_test_title}"
    }

    products_dataframe: DataFrame = spark.createDataFrame(
        unit_test_cases[unit_test_title]["Products"],
        products_table_schema
    )

    orders_dataframe: DataFrame = spark.createDataFrame(
        unit_test_cases[unit_test_title]["Orders"],
        orders_table_schema
    )

    expected_result: List = sorted(unit_test_cases[unit_test_title]["Output"])


    # 🚀 Execute every approach
    for approach_name, approach_function in approaches:

        # ⏱️ Start timer
        start_time = time.perf_counter()

        sql_result_dataframe, dataframe_result = approach_function(
            products_dataframe,
            orders_dataframe
        )

        # ⏱️ Stop timer
        end_time = time.perf_counter()

        execution_time = end_time - start_time


        # 🔍 Convert Spark output to Python objects
        sql_output = sorted(
            [
                list(row.asDict().values())
                for row in sql_result_dataframe.collect()
            ]
        )

        dataframe_output = sorted(
            [
                list(row.asDict().values())
                for row in dataframe_result.collect()
            ]
        )


        # ✅ Validate results
        sql_passed = sql_output == expected_result
        dataframe_passed = dataframe_output == expected_result

        overall_passed = sql_passed and dataframe_passed


        # 📊 Store result with execution time
        test_case_result[approach_name] = (
            f"{'🎉 PASS' if overall_passed else '❌ FAIL'} "
            f"({execution_time:.4f}s)"
        )


    overall_summary.append(test_case_result)



# 📊 Print final matrix
print("\n" + "=" * 140)
print("📊 Test Cases vs Approaches Performance Summary")
print("=" * 140)


print(
    f"{'Test Case':50}"
    f"{'🟣 Pattern Matching':30}"
    f"{'🟢 BETWEEN Dates':30}"
    f"{'🟠 YEAR & MONTH Functions':35}"
)

print("-" * 140)


for result in overall_summary:
    print(
        f"{result['Test Case']:50}"
        f"{result['🟣 Pattern Matching']:30}"
        f"{result['🟢 BETWEEN Dates']:30}"
        f"{result['🟠 YEAR & MONTH Functions']:35}"
    )


print("=" * 140)
````
