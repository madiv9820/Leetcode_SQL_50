# [📊 Summarizing Products Sold Each Day](https://leetcode.com/problems/group-sold-products-by-the-date/description/?envType=study-plan-v2&envId=top-sql-50)
### 🛍️ Story Time — The Marketplace Manager
Every evening, the manager of a bustling marketplace walks through the day's sales records before closing the shop. 🌇📋

Throughout the day, hundreds of customers buy all kinds of products. Sometimes multiple customers purchase the **same product**, while other times they buy completely different items. 🛒✨

The sales log faithfully records **every single transaction**, but the manager isn't interested in how many times each product was sold. Instead, they want a neat daily report answering two simple questions:

- **🧮 How many different products were sold today?**
- **📝 Which products were they?**

To make the report easy to read, the manager also insists that:

🔤 Product names must appear in **alphabetical (lexicographical) order**.
📅 Every day's report should be listed from the **earliest date to the latest**.

Your job is to transform the raw sales log into this clean and organized daily summary. 📊
> **💡 Remember:** If the same product is sold multiple times on the same date, it is still considered **one unique product** for that day's report.

### 🎯 Your Goal
For **each selling date**, determine:
- 📦 The number of **distinct products** sold.
- 📝 A comma-separated list of those unique product names.
- 🔤 The product names should be sorted alphabetically.
- 📅 Return the results ordered by **`sell_date`**.

````python
# =============================================================================
# 🚀 Create Spark Session
# =============================================================================
spark = SparkSession.builder.appName('Group Sold Products By The Date').getOrCreate()
````

````python
# =============================================================================
# 📋 Activities Table Schema
# =============================================================================
activities_table_schema: StructType = StructType([
    StructField('sell_date'	, DateType()	, True),
    StructField('product'	, StringType()	, True),
])
````

---
### 🚀 Approach: Grouping with Aggregation
#### 💡 Intuition
The goal is to create a daily sales summary where each date shows ***the number of unique products sold and a comma-separated list of those products in alphabetical order***. Since all required information is grouped by the selling date, aggregating the records for each date is the most natural and efficient approach.

#### 📝 Steps
1. Group all records by **`sell_date`**.
2. **Remove duplicate products** for each date to ensure each product is counted only once.
3. **Count** the number of distinct products sold on that date.
4. **Sort** the unique product names alphabetically.
5. **Join** the sorted product names into a comma-separated string.
6. **Order** the final result by sell_date in ascending order.

#### 📌 Pseudocode
```
Group records by sell_date

For each date:
    Collect unique product names
    Count the unique products
    Sort the product names alphabetically
    Join them into a comma-separated string

Sort the results by sell_date

Return the final summary
```

#### ⏱️ Complexity Analysis
- **Time Complexity: `O(N + G × K log K)`**
    - **`N`** = Total number of sales records.
    - **`G`** = Number of unique selling dates.
    - **`K`** = Number of distinct products sold on a particular date.
    - Grouping and collecting unique products take **`O(N)`**, while sorting the distinct products for each date contributes **`O(K log K)`**.

- **Space Complexity: `O(K)`**
    - Stores the set of unique products for each date during aggregation, where **`K`** is the maximum number of distinct products sold on a single day.

---

````python
# =============================================================================
# 📦 Group Sold Products By The Date
# =============================================================================
def group_sold_products_by_the_date(Activities: DataFrame) -> Tuple[DataFrame]:
    """
    📅 Groups products sold on each date.

    Returns:
        1️⃣ SQL solution result
        2️⃣ Spark DataFrame API solution result
    """
    
    # -------------------------------------------------------------------------
    # 🪄 Register DataFrame as a temporary SQL table
    # -------------------------------------------------------------------------
    Activities.createOrReplaceTempView('Activities')
    
    # -------------------------------------------------------------------------
    # 📝 SQL Solution
    # -------------------------------------------------------------------------
    sql_query = '''
        SELECT 	  sell_date
                , COUNT(DISTINCT(product)) 							AS nums_sold
                , CONCAT_WS(',', SORT_ARRAY(COLLECT_SET(product)))  AS products	
        FROM 		Activities
        GROUP BY 	sell_date
        ORDER BY  	sell_date
    '''

    sql_result_dataframe: DataFrame = spark.sql(sql_query)

    # -------------------------------------------------------------------------
    # ⚡ Spark DataFrame API Solution
    # -------------------------------------------------------------------------
    grouped: DataFrame = (
        Activities
            .groupBy(F.col('sell_date'))
            .agg(F.sort_array(F.collect_set(F.col('product'))).alias('product_set'))
    )
    spark_api_dataframe: DataFrame = (
        grouped
            .select(
                'sell_date',
                F.size(F.col('product_set')).alias('nums_sold'),
                F.concat_ws(',', F.col('product_set')).alias('products')
            )
            .orderBy(F.col('sell_date'))
    )

    return sql_result_dataframe, spark_api_dataframe
````
